## Vector Operations

```python
# Sketch u + w, x - y, and v - (w + u).
vectors = {
  'u': (-3, 5),
  'w': (5, -13),
  'y': (1, 4, 2),
  'v': (2, 1),
  'x': (4, 5, -7),
}

vec_to_plot_2D = {
  'u+w': (vectors['u'][0] + vectors['w'][0], vectors['u'][1] + vectors['w'][1]),
  'v-(w+u)': (
    vectors['v'][0] - (vectors['w'][0] + vectors['u'][0]),
    vectors['v'][1] - (vectors['w'][1] + vectors['u'][1]),
  ),
}

vec_to_plot_3D = {
  'x-y': (
    vectors['x'][0] - vectors['y'][0],
    vectors['x'][1] - vectors['y'][1],
    vectors['x'][2] - vectors['y'][2],
  ),
}

result_2d = plot_basic_vectors(
  vec_to_plot_2D,
  title='Vector Operations (2D): u + w, v - (w + u)',
  limits={'x': (-10, 10), 'y': (-10, 10)},
)

result_3d = plot_basic_vectors(
  vec_to_plot_3D,
  title='Vector Operation (3D): x - y',
  limits={'x': (-5, 5), 'y': (-5, 5), 'z': (-10, 0)},
)

result_2d, result_3d
```

## LINEAR SUBSPACES

```python
# Va = {(x, y) ∈ℝ2 |x, y ≤−1}
all_range = np.linspace(-10, 10, 100)
line_x = {
  'x': -1 * np.ones_like(all_range),
  'y': all_range,
  'label': '$x = -1$',
}
line_y = {
  'x': all_range,
  'y': -1 * np.ones_like(all_range),
  'label': '$y = -1$',
}
fill = {
  'x': all_range,
  'y1': -10 * np.ones_like(all_range),
  'y2': -1 * np.ones_like(all_range),
  'where': all_range <= -1,
}
title = '$V_a = \\{(x, y) \\in \\mathbb{R}^2 \\mid x \\leq -1, y \\leq -1\\}$'

plot_lines(
  title,
  [line_x, line_y],
  [fill],
  limits={'x': (-10, 2), 'y': (-10, 2)},
)
```

```python
# Vb = {(x, y) ∈ℝ2 |y < x2}
all_range = np.linspace(-10, 10, 100)
line = { 'x': all_range, 'y': all_range**2, 'label': '$y = x^2$' }
fill = {
  'x': all_range,
	'y1': -10 * np.ones_like(all_range),
	'y2': all_range**2,
	'where': all_range**2 > -10, # This condition is always true, so it fills the area below the parabola.
}
title = '$V_b = \\{(x, y) \\in \\mathbb{R}^2 \\mid y < x^2\\}$'

plot_lines(title, [line], [fill], limits={'x': (-10, 10), 'y': (-10, 10)})
```

```python
# Vc = {(x, y) ∈ℝ2 |x, y ≥0,y ≤x}
all_range = np.linspace(-10, 10, 100)
line_1 = {
	'x': all_range,
	'y': all_range,
	'label': '$y = x$'
}
fill_1 = {
	'x': all_range,
	'y1': np.zeros_like(all_range),
	'y2': all_range,
	'where': all_range >= 0
}
line_2 = {
	'x': all_range,
	'y': np.zeros_like(all_range),
	'label': '$y = 0$'
}
fill_2 = {
	'x': all_range,
	'y1': np.zeros_like(all_range),
	'y2': all_range,
	'where': all_range >= 0
}
title = '$V_c = \\{(x, y) \\in \\mathbb{R}^2 \\mid x \\geq 0, y \\geq 0, y \\leq x\\}$'
plot_lines(
	title,
	[line_1, line_2],
	[fill_1, fill_2],
	limits={'x': (-1, 10), 'y': (-1, 10)},
)
```

## SPANS AS SUBSPACES

```python
# V = Span([1, −3])
# V = Span([0, 2])
vectors = {
	'v1': (1, -3),
	'v2': (0, 2),
}

result = plot_basic_vectors(
	vectors,
	title='Vectors and line defined by v1 and v2',
	limits={'x': (-6, 6), 'y': (-6, 6)},
	new_figure=True,
	setup_axes=True,
	show=False,
 )

v1 = np.array(result['v1'], dtype=float)
v2 = np.array(result['v2'], dtype=float)
direction = v2 - v1

if np.allclose(direction, 0):
	raise ValueError('v1 and v2 must be different points to define a line')

# Parametric line through v1 and v2: p(t) = v1 + t * (v2 - v1)
t = np.linspace(-10, 10, 400)
line = {
	'x': v1[0] + t * direction[0],
	'y': v1[1] + t * direction[1],
	'label': 'Line through v1 and v2',
	'color': 'tab:green',
	'linestyle': '--',
}

plot_lines(
	'Vectors and line defined by v1 and v2',
	[line],
	limits={'x': (-6, 6), 'y': (-6, 6)},
	new_figure=False,
	setup_axes=False,
	show=True,
 )
```

```python
# y=3x+2 with x=-1,0,1
# V = Span([-1, -1])
# V = Span([0, 2])
# V = Span([1, 5])
vectors = {
    'v1': (-1, -1),
    'v2': (0, 2),
    'v3': (1, 5),
}

limits = {'x': (-8, 8), 'y': (-8, 8)}
result = plot_basic_vectors(
    vectors,
    title='Vectors and One Line Through v1, v2, v3',
    limits=limits,
    new_figure=True,
    setup_axes=True,
    show=False,
 )

v1 = np.array(result['v1'], dtype=float)
v2 = np.array(result['v2'], dtype=float)
v3 = np.array(result['v3'], dtype=float)

direction = v2 - v1
if np.allclose(direction, 0):
    raise ValueError('v1 and v2 must be different points to define a line')

# Check whether v3 is on the same line as v1 and v2.
cross = direction[0] * (v3 - v1)[1] - direction[1] * (v3 - v1)[0]
if not np.isclose(cross, 0):
    raise ValueError('v1, v2, and v3 are not collinear; cannot draw one shared line')

t = np.linspace(-10, 10, 400)
line = {
    'x': v1[0] + t * direction[0],
    'y': v1[1] + t * direction[1],
    'label': 'One line through y=3x+2',
    'color': 'tab:green',
    'linestyle': '--',
}

plot_lines(
    'Vectors and One Line Through y=3x+2',
    [line],
    limits=limits,
    new_figure=False,
    setup_axes=False,
    show=True,
 )
```
